Fantasy Pros Projections Web Scraper

```python
import bs4
from bs4 import BeautifulSoup 
import pandas as pd
import requests
```

```python
Positions = ["QB","RB","WR","TE"]

for position in Positions:
    #QBs don't have ppr scoring
    if position == 'QB':
        url = 'https://www.fantasypros.com/nfl/projections/qb.php?week=draft'
    else:
        url = f'https://www.fantasypros.com/nfl/projections/{position.lower()}.php?week=draft&scoring=PPR'
        
    response = requests.get(url)
    print(response)
    
    # Lets take a look at the response
    # open a file to write the response content
    # the w in open means write mode
    # f.write means we are writing to the file 
    # The text attribute of the response object contains the HTML content
    with open("response.html", "w", encoding="utf=8") as f:
        f.write(response.text)
            
    soup = BeautifulSoup(response.content, 'html.parser')
    print(soup.prettify())
    
    results = soup.find_all('caption', class_ = 'hidden-aria')
    
    for result in results:
        print(result.text) 
    
    
```
